```python
from qiskit import QuantumCircuit

qc = QuantumCircuit(3)

qc.h([0, 1, 2])
qc.x(2)
qc.h(2)
qc.mcx([0, 1], 2)
qc.h(2)
qc.x(2)

qc.draw(
    output="mpl",
    style="iqp",
    fold=-1,
    filename="circuit.png"
)
```
